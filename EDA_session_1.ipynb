- what happened is that your CSV file isn’t comma‑separated — it’s semicolon‑separated. 
    - By default, pd.read_csv() assumes commas as the delimiter. 
        - Since your file uses ;, pandas treated the entire line as one big string in a single column.
        - when we wirte sep is ; then it give proeprs
    - CSV means Comma Separated Values, so the default assumption is that commas are the delimiter.

But in practice, many datasets (especially from Europe or certain software exports) use semicolons ; instead of commas. That’s why pandas gives you one giant string column if you don’t specify sep=";"
