**bar chart**

- x-axis: class names(keys) 
- y-axis: Frequency(values)

```python
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as  sns 

```

```python
keys = visa_df['continent'].value_counts().keys()
values = visa_df['continent'].value_counts().values
```

```python

plt.figure(figsize=(10,4))   #  x is horizonatl and y is vertical
plt.bar (keys,values)
```

```python
colors=['Red','Blue','yellow','pink','Black']
plt.figure(figsize=(10,4))   #  x is horizonatl and y is vertical
plt.bar (keys,values,color=colors)
plt.xlabel('Continents')
plt.ylabel('No of applicants')
plt.title('Bar chart')
plt.show()
```

```python
***pie chart***
```

```python
keys = visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
#plt.pie(x,labels) x= where values lables - colunbs name 
plt.pie(values, 
        explode=[0.1 for i in range(6)],# gve cutss of cut part
        labels=keys,
       autopct='%1.2f%%',
       startangle=90)  # rountation off pie chart
plt.savefig('Continent_pie_jpg')
plt.show()
```

autopct = automaticpercentage 
autopct = %a.bf%%
a: how much width 
b: after point ho much decimals 
-&&: % symbl 
- %1:1f:1:40
-2:2f : 30:67%%
