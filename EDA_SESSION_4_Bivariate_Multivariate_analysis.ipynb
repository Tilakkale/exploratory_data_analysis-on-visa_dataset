```python
keys = visa_df['continent'].value_counts().keys()
clist,dlist=[],[]
for i in keys:
    cond1 = visa_df['continent']=='Asia'
    cond2 = visa_df['case_status']=='Certified'
    np.sum(cond1&cond2)
    certi_count=np.sum(cond1&cond2)
    clist.append(certi_count)
    
    cond1 = visa_df['continent']=='Asia'
    cond3 = visa_df['case_status']=='Denied'
    np.sum(cond1&cond3)
    denid_count=np.sum(cond1&cond3)
    dlist.append(denid_count)

pd.DataFrame(zip(clist,dlist),columns=['Certified','Denied'],
index=keys)
```

```python
# here we used i for every continent count for i for every kys
keys = visa_df['continent'].value_counts().keys()
clist,dlist=[],[]
for i in keys:
    cond1 = visa_df['continent']==i
    cond2 = visa_df['case_status']=='Certified'
    np.sum(cond1&cond2)
    certi_count=np.sum(cond1&cond2)
    clist.append(certi_count)
    
    cond1 = visa_df['continent']==i
    cond3 = visa_df['case_status']=='Denied'
    np.sum(cond1&cond3)
    denid_count=np.sum(cond1&cond3)
    dlist.append(denid_count)

pd.DataFrame(zip(clist,dlist),columns=['Certified','Denied'],
index=keys)
```
