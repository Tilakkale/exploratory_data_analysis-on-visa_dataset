```python
# Random fill
df1.fillna(30)  # it ranodmly filll value where null is given 
it does not “randomly” fill missing values. Instead, it systematically replaces every NaN (null) in the DataFrame with the constant value 30.

🔎 Key Behavior
fillna(value) → replaces all missing entries with the exact value you provide.

In your case, every NaN becomes 30.

It’s deterministic, not random — the same missing cell will always get 30.
    all columns give filling not by columns iwse every colums ike names and age asalry not good 
```


**pad,ffill is same , bfill  backfill is same** 
- im not doing inplace=True

```python
In older pandas (≤1.x), you could write df.fillna(method='ffill') or df.fillna(method='bfill').

In pandas 3.0+, they simplified the API:

fillna() → only for filling with a specific value (e.g., df.fillna(0)).

ffill() / bfill() → for propagating values forward/backward.


    # You can also specify axis
df1.ffill(axis=0)   # fill down the rows
df1.ffill(axis=1)   # fill across the columns
df1.bfill(axis=0)   # fill down the rows
df1.bfill(axis=1)   # fill across the columns
🔎 Quick Recap
Forward fill (ffill) → carries the last valid value forward.

Backward fill (bfill) → fills missing values using the next valid value.

fillna(value) → replaces missing values with a constant (like 30).
```

```python
# pad,ffill,bfill,backfill
# axis=0  axis=1
df1.fillna(method='ffill')
df1.fillna(method='ffill',axis=1)

df1.fillna(method='bfill')
df1.fillna(method='bfill',axis=1)
```

**KNNImputer**

- KNN : K-nearest neighbours

- where K is a number

- instaed of taking avg of all the values data

    - first will find the neighbours who are close with missing data 

    - by using Euclidian distance calculation will idenify the neighbour rows

    - suppose k=2 will identify the two closest rows

    - those closest rows values will take it, will calculate the mean

    - finally will fill with the mean

```python
salary=[50000,np.nan,70000,90000]
age=[25,28,30,40]
df=pd.DataFrame(zip(salary,age),
            columns=['Salary','Age'])
df
```

```python
from sklearn.impute import KNNImputer 
imputer=KNNImputer(n_neighbors=2)
pd.DataFrame(imputer.fit_transform(df))
```

```python
Imputation = the process of filling in missing values in a dataset.

Instead of dropping rows/columns with NaN, we replace them with something meaningful.

Common strategies:

Constant value (e.g., fillna(0) or fillna("Unknown"))

Statistical value (mean, median, mode of the column)

Forward/backward fill (ffill, bfill)

Model-based imputation (predict missing values using other features)

⚡ KNN Imputer
KNNImputer is a specific imputation method from sklearn.impute.

It uses the K-Nearest Neighbors algorithm to fill missing values.

Steps:

For a row with a missing value, find its k nearest rows (neighbors) based on other feature values.

Replace the missing value with the average (or weighted average) of those neighbors’ values.

This is smarter than just filling with mean/median because it considers the similarity between rows.
```
